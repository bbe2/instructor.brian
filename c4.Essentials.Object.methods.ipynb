**`Example: Plain english: Object, Instance, and Method`**

**`Class`**  
-> A class is like a blueprint or recipe  
-> It describes how a certain kind of thing should look and what it can do, but it isn’t the thing itself  

**`Instance (object)`**  
-> When you use a class to build something real, you create an instance (also called an object)  
-> Each instance gets its own data — like a unique copy built from the same plan  

**`Method`**   
-> Methods live as a function inside a class   
-> It’s a set of actions the object knows how to do  
-> When you call a method on an object, Python passes the object itself as the first input (called `self`)  
-> enables a method to work with its object’s data as object.method()  


````python
# ====== EXAMPLE: class vs object in Python ======
' define a class: blueprint for making objects'

class Dog:                      # __init__ is the constructor; it runs when we make a new Dog
  def __init__(self, name):   # 'self' is the new object being built; we attach data to it
    self.name = name
  def bark(self):            # define a method (behavior) that every Dog can do
    return f"{self.name} says woof"

# ---- create an object (instance) ----
  ' >Dog< is the class (the plan)'
  ' >fido< is an object built from that plan'

fido = Dog("Fido")
print(isinstance(fido, Dog))   # True, # test that fido is really a Dog object
print(fido.bark())             # "Fido says woof", # call fido's method

# inspect the class of fido
print(type(fido))              # <class '__main__.Dog'>
print(type(Dog))               # <class 'type'> (Dog itself is an object too!)

# ---- summary notes ----
# class  -> defines structure and behavior (attributes + methods)
'''classes are objects, ie an instances of 'type' w real data'''

````

````python
# Musical Text-to-Rhythm (Beautiful + Text-Connected Melody)
# ----------------------------------------------------------------------

# Step 0: Install missing package if needed
# !pip install pyphen  # Uncomment in Colab if not yet installed

!pip install pyphen
import pyphen
import numpy as np
import matplotlib.pyplot as plt
import IPython.display as ipd
import wave, struct
import nltk

nltk.download('punkt')

# Step 1: Input poem (structured by line)
lines = [
    "Gretel knocked upon the door",
    "candy walls and cookie floor",
    "witch inside with eyes aglow",
    "come in children don't be slow"
]

# Step 2: Break lines into syllables
syllabifier = pyphen.Pyphen(lang='en')

def parse_syllables_by_line(lines):
    result = []
    for line in lines:
        line_sylls = []
        for word in line.split():
            sylls = syllabifier.inserted(word).split("-")
            for s in sylls:
                line_sylls.append((s, word))
        result.append(line_sylls)
    return result

structured = parse_syllables_by_line(lines)
all_sylls = [s for line in structured for s, _ in line]

# Step 3: Display visual rhythm
plt.figure(figsize=(len(all_sylls), 2))
for i, syl in enumerate(all_sylls):
    plt.text(i+0.5, 0.7, '♩', ha='center', fontsize=14)
    plt.text(i+0.5, 0.3, syl, ha='center', fontsize=12)
plt.axis('off')
plt.show()

# Step 4: Melody mapping (rising/falling within line)
c_major = [523, 587, 659, 698, 784, 880, 988, 1046]

def phrase_to_melody(phrase, index):
    count = len(phrase)
    mid = count // 2
    scale = np.linspace(c_major[0], c_major[min(count - 1, len(c_major)-1)], count)
    melody = []
    for i, (syll, _) in enumerate(phrase):
        freq = scale[i] if i < mid else scale[count - i - 1]
        melody.append((freq, syll))
    return melody

melody = []
for i, line in enumerate(structured):
    melody.extend(phrase_to_melody(line, i))

# Step 5: Audio synthesis

def generate_audio(melody, bpm=90, filename="melody.wav"):
    fs = 44100
    dur = 60 / bpm
    amp = 14000
    samples = []

    for freq, syll in melody:
        for t in np.linspace(0, dur * 0.8, int(fs * dur * 0.8), False):
            sample = int(amp * np.sin(2 * np.pi * freq * t))
            samples.append(sample)
        samples.extend([0] * int(fs * dur * 0.2))

    with wave.open(filename, 'w') as wf:
        wf.setnchannels(1)
        wf.setsampwidth(2)
        wf.setframerate(fs)
        wf.writeframes(b''.join(struct.pack('<h', s) for s in samples))

    return filename

# Step 6: Playback
final_audio = generate_audio(melody)
ipd.display(ipd.Audio(final_audio))
````
