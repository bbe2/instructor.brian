## **`>_13.1 Visualization - Mental model of streams, files, and formats`**


- Programming languages give you **streams** and **file handles** where bytes physically live
- At runtime, a file is opened and read/write via bytes or lines through a high-level API
-> consider these options

```
Java
- InputStream / Reader    → read bytes or characters from a source
- OutputStream / Writer   → write bytes or characters to a destination
- try-with-resources      → ensures streams are closed

Python
- open(path, mode)        → returns a file object (stream)
- file.read(), file.write()  → read/write text or bytes
- with open(...) as f:    → ensures the file is closed

Idea map
- SOURCE → (stream abstraction) → PROGRAM → (stream abstraction) → SINK (see data)
- Key idea - a file is one kind of source/sink
-> others patterns include: network sockets, stdin/stdout and compressed archives
```

**`formats: plain text vs structured text`**

- Plain text files like logs, and `.txt` are lines of strings
-> can be read line by line, tokenized and similar
- Structured formats like **XML** and **JSON** add predictable syntax around data
-> parsers turn text into trees and more commonly nested dictionaries

- - XML: nested tags `<host>`, `<port>` with attributes  
- - JSON: `{ "ip": "192.168.0.10", "ports": [22, 80] }`

### **`13.3.1: safe parsing skeleton with defusedxml + lxml`**

- Security First: It swaps the standard parser for defusedxml to automatically block "XML bombs" or malicious files happy to crash your program

- Object Creation: It converts the raw text block into a Tree Object (root) that Python understands and navigates

- Outcome: The print output <Element nmaprun...> confirms the structure is valid and ready for data extraction

````
# 13.3.1 safe parsing skeleton with defusedxml + lxml

from defusedxml.lxml import fromstring  # safe wrapper
from lxml import etree

# For the lab, you'll typically read this from a file. Here we keep a tiny snippet inline.
nmap_xml_sample = b"""
<nmaprun>
  <host>
    <address addr=\"192.168.0.10\" addrtype=\"ipv4\" />
    <ports>
      <port protocol=\"tcp\" portid=\"22\">
        <state state=\"open\" />
      </port>
      <port protocol=\"tcp\" portid=\"80\">
        <state state=\"closed\" />
      </port>
    </ports>
  </host>
</nmaprun>
"""

# parse the XML into a safe tree
root = fromstring(nmap_xml_sample)

print("Root tag:", root.tag)
print(root)

````

### **`13.3.2: extract hosts and open ports from parsed XML`**




```
This code teaches the essential process of data extraction and
- structural transformation from a hierarchical format (XML)
- into a flat Python data object like lists or dictionaries
```

1. **`Iterative Walking`** The code uses two nested loops (findall) to walk the XML tree.
-> First it finds every <host>, then inside each host, it finds every relevant <port>

2. **`Conditional Filtering`** It applies a rule (state_elem.get("state") == "open") to throw away most of the data.
-> Only ports marked "open" are collected; all others are ignored

3. **`Data Reshaping`** It transforms the complex XML structure into a Python dictionary list
-> [{"ip": ..., "open_tcp_ports": [...]}]
-> making the network data simple to read, process, and use in other programs

````
# 13.3.2: extract hosts and open ports from parsed XML

def extract_hosts(root):
    """
    Walk the XML tree and return a list of dicts:
    [{"ip": "x.x.x.x", "open_tcp_ports": [22, 80, ...]}, ...]
    """
    results = []

    for host in root.findall("host"):
        addr_elem = host.find("address")
        if addr_elem is None:
            continue

        ip = addr_elem.get("addr")
        ports_elem = host.find("ports")
        open_ports = []

        if ports_elem is not None:
            for port in ports_elem.findall("port"):
                state_elem = port.find("state")
                if state_elem is None:
                    continue

                if state_elem.get("state") == "open":
                    # portid attribute is a string; convert to int if desired
                    portid = int(port.get("portid", 0))
                    open_ports.append(portid)

        results.append({
            "ip": ip,
            "open_tcp_ports": open_ports,
        })

    return results

host_summaries = extract_hosts(root)
host_summaries
````

### **`_Lab 13.2 — safe XML parsing and host extraction`**

**Goal:** Parse the nmap XML safely and extract a summary of hosts and open ports

```
Tasks:

```

> Stretch: also record the number of closed ports or the overall scan state for each host


````
# 13.3.2: extract hosts and open ports from parsed XML

def extract_hosts(root):
    """
    Walk the XML tree and return a list of dicts:
    [{"ip": "x.x.x.x", "open_tcp_ports": [22, 80, ...]}, ...]
    """
    results = []

    # Using the safer method: XPath expression to find all 'host' elements directly under the root
    for host in root.findall(".//host"): # Use .// to search deeply if needed, though 'host' is often direct
        addr_elem = host.find("address")
        if addr_elem is None:
            continue

        ip = addr_elem.get("addr")
        ports_elem = host.find("ports")
        open_ports = []

        if ports_elem is not None:
            # Find all 'port' elements under 'ports'
            for port in ports_elem.findall("port"):
                state_elem = port.find("state")
                if state_elem is None:
                    continue

                if state_elem.get("state") == "open":
                    # portid attribute is a string; convert to int
                    # Use .get() with a default for robustness
                    portid_str = port.get("portid")
                    if portid_str and portid_str.isdigit():
                         open_ports.append(int(portid_str))

        results.append({
            "ip": ip,
            "open_tcp_ports": open_ports,
        })

    return results

host_summaries_sample = extract_hosts(root_sample)
print("\nSample Host Summaries:")
print(host_summaries_sample)
````
