### **`_Normalization`**

**`Normalization Definition & Learning Tree`**

**`▼ >_1.0 ROOT DEFINITION`**
```
  │   ├─> The process of restructuring a relational database in accordance
  │   │     with a series of "normal form" rules in order to reduce data
  │   │     redundancy and improve data integrity
  │   │
  │   └─> Plain English: Organizing a messy room!
  |         Instead of throwing everything like socks, books, pants into one
  │         giant box, you build specific drawers for specific items to help
  │         find, add, or exchange them easily
  ▼

  
→ Key Concept => "Single Source of Truth" (SSOT)
   - Normalization ensures that every piece of data, like a book title or
      patron address, lives in exactly one "home" database table
  
  - This prevents data conflicts where System A thinks the user lives
     in Boston but System B thinks they live in New York

```

**`▼ >_2.0 WHY DO WE DO IT?`**

```
  ├── Driver A: Redundancy
  │   │
  │   ├─> Concept: quit storing same fact in multiple places
  │   └─> Plain English:
  |          -> If a user changes their email address it should only be
  |             changed one time, in one place versus 500 different rows
  │
  ├── Driver B: Anomalies
  │   │
  │   ├─> Insertion Anomaly: Cannot add data because other data is missing
  │   ├─> Deletion Anomaly: Deleting data unintentionally wipes out other facts
  │   └─> Plain English:
  |         -> Dont create fake 'placeholder' but directly to the system
  ▼
```

**`▼ >_3.0 THE LEARNING HIERARCHY`**

```
  ┌─── STEP 1: SHAPE - 1st normal form  ────────────────────────┐
  |                                                             |  
  │  1NF (First Normal Form) -> "The Grid"                      │
  │  Goal: Atomicity or one value per cell                      │
  │   -> Plain English: No lists inside a single box            │
  └──────────────────────────────┬──────────────────────────────┘
                                 ▼
  ┌─── STEP 2: GLUE - 2nd normal form ──────────────────────────┐
  |                                                             |
  │  2NF (Second Normal Form) -> "is type of separation"        │
  │  Goal: Remove Partial Dependency                            │
  │   -> Plain English: Facts are independent                   │
  └──────────────────────────────┬──────────────────────────────┘
                                 ▼
  ┌─── STEP 3: PURITY - 3rd normal form ────────────────────────┐
  |                                                             |  
  │  3NF (Third Normal Form) -> "The Reference"                 │
  │  Goal: Remove Transitive Dependency                         │
  │   -> Plain English: Eliminate the middle-man                │
  └─────────────────────────────────────────────────────────────┘

```

````

▼ >_2.0 First Normal Form (1NF)
  │
  ├─> ┌─── The long table ───────────────────────┐
  │   │ 4844124 | Rusty    | Book (Toyota Camry) |
  │   │ 4844124 | Lucyanna | Movie (Rush)        |
  │   └──────────────────────────────────────────┘
  │
  ├─> ✓ FIXED: Every cell holds only one value.
  ├─> ⚠ PROBLEMS: Massive Redundancy.
  |         -> "Rusty" and "4844124" appear repeatedly
  │             for every book he checks out
  │
  └─> Plain English:
       -> Unstake the list so every item has its own line but we're
          repeating the Patron's name over and over again
          -> safe but wasteful!

  ▼ (Action: Remove Partial Dependencies)
  │ "Does 'Book Title' depend on the Patron? No. Split it."
  ▼
````

````

▼ >_3.0 Second Normal Form (2NF)
  │
  ├── [Table A: Loan Info]       ├── [Table B: Item Info]
  │   (Links Patron to Item)         (Static facts about books/movies)
  │
  ├─> ✓ FIXED: Data about the *Item* (Title, Format) is separated from
  │   the act of *loaning* it
  │
  └─> Plain English: Separate the 'Event' from the 'Object'. The fact that
      'Rush' is a DVD doesn't change just because Rusty borrowed it
         -> in short, "movies" get their own home

      ▼ (Action: Remove Transitive Dependencies)
      │ "Does 'PatronLastName' depend on the PatronCard? Yes."
      │ "Does it depend on the Loan? No. Move it further out."
      ▼
````

````

▼ >_4.0 Third Normal Form (3NF) - The Gold Standard
  │
  ┌──────────────┐    ┌──────────────┐    ┌──────────────┐
  │   Patrons    │    │    Loans     │    │    Items     │
  └──────┬───────┘    └──────┬───────┘    └──────┬───────┘
         │                   │                   │
  (ID, Name, Addr)    (Date, Due)         (Title, Type, Author)
  │
  └─> Plain English:
        -> Every piece of data lives in the one place that defines it
      - The Patron's name lives with the Patron ID
      - The Book's author lives with the Book ID
      - The Due Date lives with the Loan transaction
      - Nothing is "piggybacking" on the wrong ID
````

### **`_Lesson 12.2 - Detail overview`**

````
12.1 Databases overview — tables, queries, and relationships
by: Brian Hogan, brian.hogan@regiscollege.edu

This lesson is about using a relational database to turn raw data into structured
tables for performing data queries on.
  - instead of scattered lists, ad hoc dicts, or many csv files, you design a
    schema made of tables with named columns and constraints
  - each table stores one kind of thing (students, courses, enrollments), and SQL
    describes *what* data you want from those tables, while sqlite handles *how*
    to find it on disk

The core pattern starts with a table with clearly defined columns and a primary
key that uniquely identifies each row.
  - related tables connect through keys, so a student row can link to many
    enrollment rows, and each enrollment row can link to a course
  - select queries pull rows from one table, and joins combine rows across
    tables so you can answer questions like “which students are in which course”
    or “how“ many credits is this student taking.” where filters narrow results,
    and order by controls how they are sorted

Python pulls data together with an SQLite script to the .db file, sending SL
instructions like SELECT name, gpa FROM students WHERE gpa >= 3.0, and receives
rows back as tuples or dataframes data objects.
  - the same program code can run many different queries against the same schema,
    or reuse the same query on different databases
  - pandas read_sql_query lets you treat query results as dataframes, making it
    easy to filter, aggregate, and visualize database data in colab

Lab 12 is working with a small schema and performing the queries.
  - the enrollment lab has you create students, courses, and enrollments tables,
    enforce a simple primary key, and write select queries that join tables to
    answer real questions
  - a final step brings the data into pandas to sort, count, and summarize results
  - the important shift is to see tables, keys, and queries as a shared language
    between the python programs and gathered data, not just as syntax for storing rows


Database Keywords:
================
Name / Pattern                      | Category         | Description
------------------------------------|------------------|---------------------------------------------------------=
`table`                             | core structure   | collection of rows with named columns
`row` / `record`                    | data unit        | one item stored in a table (one student, one course)
`column` / `field`                  | attribute        | named property such as name, gpa, or course_id
`schema`                            | design blueprint | set of tables, columns, and relationships in a database
`primary key`                       | constraint       | column (or combo) that uniquely identifies each row
`foreign key`                       | relationship     | column that refers to a primary key in another table
`sqlite`                            | database engine  | lightweight relational database used in a single .db file
`school.db` (or similar)            | .db file         | on-disk file holding all tables and data for this lesson
`SQL` (Structured Query Language)   | query language   | describes what data to select, insert, update, or delete
`SELECT ... FROM ...`               | query pattern    | basic pattern for reading rows from one or more tables
`WHERE`                             | filter clause    | narrows results based on conditions (gpa >= 3.0, etc.)
`JOIN`                              | table connector  | combines rows from related tables using key relationships
`ORDER BY`                          | sort clause      | controls ordering of rows in query results
`sqlite3.connect()`                 | python api       | opens a connection to a sqlite .db file
`cursor.execute()`                  | python api       | sends a sql command from python to the sqlite engine
`cursor.fetchall()`                 | python api       | retrieves all result rows from last executed query
`pandas.read_sql_query()`           | pandas bridge    | runs a query and returns results as a dataframe
`schema diagram` (students/courses) | design tool      | visual map of tables, keys, and relationships
`enrollment lab`                    | hands-on task    | build and query the tiny school database in colab
````

````

# 12.2.2: Creating a simple students table
# This table stores one row per student
create_students_sql = """
CREATE TABLE IF NOT EXISTS students (
    student_id INTEGER PRIMARY KEY,
    name       TEXT NOT NULL,
    major      TEXT,
    gpa        REAL
);
"""
cur.execute(create_students_sql)
conn.commit()

print("Created students table (if it did not already exist).")
````

````

# 12.2.3: Inserting sample rows using parameterized queries

students_data = [
    ("Alice", "Computer Science", 3.8),
    ("Bob",   "Data Science",     3.2),
    ("Cara",  "Mathematics",      3.5),
]

insert_sql = "INSERT INTO students (name, major, gpa) VALUES (?, ?, ?);"

# Parameterized inserts help avoid SQL injection and string formatting bugs.
cur.executemany(insert_sql, students_data)
conn.commit()

print("Inserted sample students (if this cell is run multiple times, you'll add duplicates).")
print(students_data)
````

````

# 12.2.4: Selecting rows from the students table

select_sql = "SELECT student_id, name, major, gpa FROM students;"
cur.execute(select_sql)

rows = cur.fetchall()

print("All students:")
for row in rows:
    print(row)
````

````

# Lab.12.Part.1.2: TODO — Use pandas to analyze joined enrollment data

import pandas as pd

join_sql = """
SELECT s.student_id,
       s.name,
       c.course_id,
       c.title,
       e.grade
FROM students AS s
JOIN enrollments AS e ON s.student_id = e.student_id
JOIN courses     AS c ON c.course_id = e.course_id;
"""

df_enrollments = pd.read_sql_query(join_sql, conn)
df_enrollments
````
