````python
# Smiley RL Poster: help the face reach (0,0) and level angle θ≈0
# ε-greedy bandit on simple actions; student clicks Better/Worse/Skip to reinforce
# Saves audit CSV with timestamps for async accountability

import math, os, csv, time, random
from datetime import datetime
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Arc
from ipywidgets import VBox, HBox, Button, HTML, Output

# ---- config ----
SEED = 7
random.seed(SEED); np.random.seed(SEED)

EPSILON     = 0.15         # exploration rate
MAX_TRIALS  = 50          # interaction budget
STEP_POS    = 0.6          # delta for x,y
STEP_ANG    = math.radians(8)  # delta for angle
FACE_R      = 1.1          # face radius
LOG_PATH    = "smiley_rl_poster_log.csv"

# actions: (Δx, Δy, Δθ)
ACTIONS = [
    ("left",   (-STEP_POS, 0.0,          0.0)),
    ("right",  ( STEP_POS, 0.0,          0.0)),
    ("up",     ( 0.0,      STEP_POS,     0.0)),
    ("down",   ( 0.0,     -STEP_POS,     0.0)),
    ("ccw",    ( 0.0,      0.0,   +STEP_ANG)),
    ("cw",     ( 0.0,      0.0,   -STEP_ANG)),
    ("nudge+", (+STEP_POS/2, +STEP_POS/2, -STEP_ANG/2)),
    ("nudge-", (-STEP_POS/2, -STEP_POS/2, +STEP_ANG/2)),
]

# ---- bandit state ----
K = len(ACTIONS)
Q = np.zeros(K)           # value estimates
N = np.zeros(K, dtype=int)

# ---- environment state ----
x, y, theta = 3.0, -2.8, math.radians(22)   # start off-center & tilted
trial = 0

# ---- plotting setup ----
out = Output()
status = HTML()
fig, ax = plt.subplots(figsize=(5,5))
plt.close(fig)

def draw_smiley(cx, cy, ang):
    ax.clear()
    ax.set_aspect('equal')
    ax.set_xlim(-6, 6); ax.set_ylim(-6, 6)
    ax.grid(True, linewidth=0.6, alpha=0.4)
    ax.axhline(0, linewidth=1.0, alpha=0.6)
    ax.axvline(0, linewidth=1.0, alpha=0.6)
    ax.set_xlabel("x"); ax.set_ylabel("y")
    # face outline
    face = Circle((cx, cy), FACE_R, fill=False, linewidth=2.5)
    ax.add_patch(face)
    # rotation matrix
    R = np.array([[math.cos(ang), -math.sin(ang)],
                  [math.sin(ang),  math.cos(ang)]])
    # eyes (relative coords then rotate + translate)
    def place(px, py):
        return (R @ np.array([px, py])) + np.array([cx, cy])
    eye_dx, eye_dy = 0.45, 0.35
    for sx in (-1, +1):
        ex, ey = place(sx*eye_dx, eye_dy)
        eye = Circle((ex, ey), 0.12, fill=True)
        ax.add_patch(eye)
    # mouth arc
    m_center = place(0, -0.2)
    # mouth curvature tracks angle magnitude: closer to level → happier
    curve = 30 * (1 - min(abs(ang)/math.radians(45), 1))
    arc = Arc(m_center, 1.2, 0.7, angle=np.degrees(ang), theta1=200-curve, theta2=340+curve, linewidth=2)
    ax.add_patch(arc)
    # HUD
    ax.text(-5.8, 5.4, f"target → (0,0), θ≈0°", fontsize=9, ha='left')
    ax.text(-5.8, 4.7, f"pos=({cx:+.2f},{cy:+.2f})  θ={math.degrees(ang):+.1f}°", fontsize=9, ha='left')

def select_action():
    if (N==0).any():           # ensure each arm tried at least once
        return int(np.argmin(N))
    if np.random.rand() < EPSILON:
        return np.random.randint(K)
    return int(np.argmax(Q))

def apply_action(idx):
    global x, y, theta
    dx, dy, dth = ACTIONS[idx][1]
    x += dx; y += dy; theta += dth
    # keep within frame bounds
    x = float(np.clip(x, -5.5, 5.5))
    y = float(np.clip(y, -5.5, 5.5))
    theta = float(((theta + math.pi) % (2*math.pi)) - math.pi)  # wrap to [-π,π]

def update_bandit(idx, reward):
    N[idx] += 1
    Q[idx] += (reward - Q[idx]) / N[idx]

def distance_to_goal():
    # Euclidean distance to (0,0) plus angle penalty (scaled)
    pos = math.hypot(x, y)
    ang = abs(theta)
    return pos + 0.6*ang

def log_event(idx, reward, note):
    new_file = not os.path.exists(LOG_PATH)
    with open(LOG_PATH, "a", newline="") as f:
        w = csv.writer(f)
        if new_file:
            w.writerow(["timestamp","trial","action","x","y","theta_deg","reward","dist","Q"] + [a[0] for a in ACTIONS])
        w.writerow([datetime.now().isoformat(), trial, ACTIONS[idx][0], round(x,3), round(y,3),
                    round(math.degrees(theta),3), reward, round(distance_to_goal(),3), round(Q[idx],3)] + [round(q,3) for q in Q])

# ---- UI wiring ----
btn_better = Button(description="Better 🙂", button_style="success")
btn_worse  = Button(description="Worse 🙁",  button_style="danger")
btn_skip   = Button(description="Skip / Explore", button_style="info")
btn_reset  = Button(description="Reset ↺", button_style="")
ui = VBox([status, out, HBox([btn_better, btn_worse, btn_skip, btn_reset])])

current_idx = None

def step():
    global trial, current_idx
    trial += 1
    current_idx = select_action()
    apply_action(current_idx)
    draw_smiley(x, y, theta)
    with out:
        out.clear_output(wait=True); display(fig)
    status.value = f"<b>Trial {trial}/{MAX_TRIALS}</b> — action: <b>{ACTIONS[current_idx][0]}</b> → click Better/Worse/Skip. target is (0,0) & level"

def finish():
    best = int(np.argmax(Q))
    status.value = f"<b>done.</b> best action so far: <b>{ACTIONS[best][0]}</b> | Q={Q.round(3)} | log → <code>{LOG_PATH}</code>"
    btn_better.disabled = btn_worse.disabled = btn_skip.disabled = True

def on_feedback(reward, note):
    update_bandit(current_idx, reward)
    log_event(current_idx, reward, note)
    if trial >= MAX_TRIALS:
        finish()
    else:
        step()

btn_better.on_click(lambda _: on_feedback(1.0, "better"))
btn_worse.on_click(lambda _: on_feedback(0.0, "worse"))
btn_skip.on_click(lambda _: on_feedback(0.5, "skip"))

def reset(_=None):
    global x, y, theta, trial, Q, N
    x, y, theta = 3.0, -2.8, math.radians(22)
    trial = 0
    Q[:] = 0.0; N[:] = 0
    btn_better.disabled = btn_worse.disabled = btn_skip.disabled = False
    step()

btn_reset.on_click(reset)

display(ui)
step()
````

````python
%%bash
set -e
cat > Main.java <<'EOF'
import java.security.SecureRandom;
import java.util.*;
public class Main {
  static final java.security.SecureRandom R = new java.security.SecureRandom();
  static String token(){ byte[] b=new byte[16]; R.nextBytes(b); return java.util.Base64.getUrlEncoder().withoutPadding().encodeToString(b); }
  public static void main(String[] args){
    String sessionId = java.util.UUID.randomUUID().toString();
    String csrf = token();
    java.util.Map<String,String> session = new java.util.HashMap<>();
    session.put("sid", sessionId);
    session.put("csrf", csrf);
    // submit
    String posted = csrf; // try changing to "bad"
    boolean valid = posted.equals(session.get("csrf"));
    if(valid) session.remove("csrf"); // one-time use
    System.out.println(valid ? "form accepted" : "invalid token");
  }
}
EOF
javac Main.java && java Main
````
