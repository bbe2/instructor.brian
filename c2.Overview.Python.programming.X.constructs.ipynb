````python
  <i.n.f.o.r.m.a.t.i.o.n>
    ├> step.1 -> ETL: extract, translate, load
    └>  parse information data into structures
    ▼
    ├> step.2 -> manipulate data with programming constructs
    ├────────────────────┼──────────────────────────────
    ├> A &               ├> B
        Data structures  │  Programming Constructs
        ├> Dataframe     │     ├> Conditionals
        ├> Lists         │     ├> Iterators
        ├> String        │     ├> Operators
        ├> Dictionary    │     └> Libraries (pypi.org)
                         │     ├> Functions
        ├> Tuple         │     ├> Variables
        ├> Set           │     ├> Class Objects
        ├> array         │        ├> Inheritance
        └> Stack/Queue   │        └> Polymorphism
                         │     └> Documentation

=> Essential Data Structures
object.name    | character code | constructor function            | outcome example
----------------|----------------|----------------------------------|-------------------------------------------
mytuple         = | (, )           |=> mytuple   = tuple(object)        | => ('string',(2+0j),[1,2,3])
mylist          = | [  ]           |=> mylist    = list(object)         | => ['bambam',"a+b=c",[1,2,3]]
mydict          = | {key:value}    |=> mydict    = dict(object)         | => {'customer':['john','doe']}
myset           = | set(object)    |=> myset     = set(object)          | => ['b','o','o'] [out]{'b','o'}
mystring        = | " " or ' '     |=> mystring  = str(object)          | => str(myset or mylist)
myjson          = | { } JSON text  |=> import json; data=json.loads(s)  | => {'x':3,'y':[1,2,3]}
myfrozenset     = | frozenset()    |=> myfset    = frozenset(object)    | => frozenset({'a','b'})
mybytes         = | b" "           |=> mybytes   = bytes(object,enc)    | => b'hello'
mybytearray     = | bytearray()    |=> mybarray  = bytearray(object)    | => bytearray(b'hello')
mycomplex       = | (a+bj)         |=> mycpx     = complex(a,b)         | => (2+5j)
mynone          = | None           |=> mynone    = type(None)()         | => None

'Advanced / Applied Data Structures'
object.name    | library/module | constructor function            | outcome example
----------------|----------------|----------------------------------|-------------------------------------------
mynamedtuple    = | collections    |=> from collections import namedtuple; Point=namedtuple('Point',['x','y']) | => Point(3,4)
mydeque         = | collections    |=> from collections import deque; dq=deque([1,2,3]) | => deque([1,2,3])
myarray         = | array          |=> from array import array; arr=array('i',[1,2,3])  | => array('i',[1,2,3])
mydataclass     = | dataclasses    |=> from dataclasses import dataclass; @dataclass class Car: make:str; year:int | => Car('Tesla',2025)
mydataframe     = | pandas         |=> import pandas as pd; df=pd.DataFrame(data)       | =>   col1 col2\n0    1    2\n1    3    4

==================================================
=> R1C1 => form of spreadsheet cell counting or indexing
=> row.1, column.2
========================================================

===================
=> R1C1 ITERATION
   - notation for interpreting spreadsheet data
======================================================================================
concept.name   | symbol code | constructor / pattern                | example outcome
---------------|-------------|--------------------------------------|-----------------------------------------
R1C1.indexing =| R(row)C(col)|=> f"R{row}C{col}" or index[(r,c)]    | => "R1C2"  → row=1, col=2
r1c1.loop     =| for r,c in  |=> for r in range(R): for c in range(C):| => iterates cell-wise, R0C0→RmaxCmax
r1c1.tuple    =| (r,c)       |=> each cell = (row_index,col_index)   | => (3,5)
r1c1.access   =| df.iat[]    |=> df.iat[row_index,col_index]         | => df.iat[0,1] = value in R1C2
'advanced
  - r1c1.lettermap=| chr(65+col) |=> Excel-like convert col→A,B,C...     | => (1,2) → 'B1'
  - r1c1.map      =| λ(row,col)  |=> cellmap={(r,c):f(r,c) for r in rows for c in cols} | => {(1,1):'A1',(1,2):'A2'}

========================================================
=> R1,C1 -> Conceptual Bridge
R1C1 maps spreadsheet cognition to Python iteration logic
- each cell is a 2D coordinate
- used in teaching loops, matrix ops, and DataFrame navigation

→ R1C1 is an intuitive visual cell representation of data transposed into data structures

========================================================
'Example of row & column iteration
========================================================
data = [
    ["A1","B1","C1"],
    ["A2","B2","C2"],
    ["A3","B3","C3"],  ]

rows = len(data)
cols = len(data[0])

for r in range(rows):
    for c in range(cols):
        print(f"R{r+1}C{c+1} => {data[r][c]}", end="  ")
    print()

''' output
R1C1 => A1  R1C2 => B1  R1C3 => C1
R2C1 => A2  R2C2 => B2  R2C3 => C2
R3C1 => A3  R3C2 => B3  R3C3 => C3   '''

→ key insight: len(data) dynamically determines table size fo reading data
→ iteration with len bridges abstract array logic to concrete grid navigation — a k
````

````python
\import random, requests
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
import plotly.express as px
import pandas as pd
from sklearn.decomposition import PCA
import plotly.graph_objects as go

# User inputs 1-5 plays for algorithm to guess play type
#=>example randomizes play selection for learning
num_plays_exclude = int(input("Enter 1-5 plays to predict play type: "))

# Randomly select test play quantity
df = pd.read_excel(requests.get( \
    "https://github.com/cosc-526/home.page/raw/main/data.M2.exercises.shakespeare.xlsx").content)
excluded_indices = random.sample(range(len(df)), num_plays_exclude)

# Prepare the dataset where features='script' and labels=<play>'type'
script = df['script']
play_type = df['type']

# Remove excluded test dataset plays from dataset
script = script.drop(excluded_indices)
play_type = play_type.drop(excluded_indices)

````

````python
'''==> Stage 2: Feature Extraction ==> Stage 2: Feature Extraction =>Stage 2'''

# Perform feature extraction on the script data
vectorizer = TfidfVectorizer()
script_features = vectorizer.fit_transform(script)

print(script_features)
````

````python
'''==> Stage 3: Machine Learning ==> Stage 3: Machine Learning ==> Stage 3:'''

# Train a Random Forest classifier using the dataset
rf = RandomForestClassifier()
rf.fit(script_features, play_type)

# Print the details of the excluded plays and their predicted play types
print("Excluded Plays (Test Data):")
excluded_plays = []  # Store excluded play details
for case_num, index in enumerate(excluded_indices, 1):
    play = df.iloc[index]
    title = play['title']
    play_type_actual = play['type']

    # Predict the play type of the excluded play
    test_features = vectorizer.transform([play['script']])
    predicted_play_type = rf.predict(test_features)[0]

    # Print the case number, play details, and the predicted play type
    print("Case:", case_num)
    print("Title:", title)
    print("Type:", play_type_actual)
    print("Predicted Type:", predicted_play_type)
    print()

    excluded_plays.append({
        'Index': index,
        'Title': title,
        'Type': play_type_actual,
        'Predicted Type': predicted_play_type
    })

# Get the actual play type of the excluded plays
actual_play_type = df.iloc[excluded_indices]['type']

# Predict the play type of the excluded plays
test_features = vectorizer.transform(df.iloc[excluded_indices]['script'])
predicted_play_type = rf.predict(test_features)

````

````python
'''==> Stage 4: Model Evaluation ==> Stage 4: Model Evaluation Stage 4:'''

# Calculate and print the accuracy on the excluded plays
test_accuracy = accuracy_score(actual_play_type, predicted_play_type)
print("Test Accuracy:", test_accuracy)

# Create a DataFrame with the play type and predicted play type
results_df = pd.DataFrame({'Type': actual_play_type, \
                           'Predicted Type': predicted_play_type})
````

````python
'''==> Stage 5: Visualization ==> Stage 5: Visualization ==> Stage 5: '''

# Apply PCA to the script features
pca = PCA(n_components=3)
script_features_pca = pca.fit_transform(script_features.toarray())

# Create a DataFrame with the PCA-transformed features and play type
pca_df = pd.DataFrame(script_features_pca, columns=['PC1', 'PC2', 'PC3'])
pca_df['Type'] = play_type.tolist()

# Create a scatter plot using Plotly Express
fig = px.scatter_3d(pca_df, x='PC1', y='PC2', z='PC3', color='Type')

# Add layout
fig.update_layout(
    title="3D Principal Component Analysis",
    scene=dict(
        xaxis=dict(title="Principal Component 1"),
        yaxis=dict(title="Principal Component 2"),
        zaxis=dict(title="Principal Component 3")
    ),
    showlegend=True
)

# Update the marker style for the predicted data points
for play in excluded_plays:
    predicted_type = play['Predicted Type']
    marker_color = 'black'
    if predicted_type == 'comedy':
        marker_color = 'black'
    elif predicted_type == 'tragedy':
        marker_color = 'blue'
    elif predicted_type == 'history':
        marker_color = 'purple'
    index = play['Index']
    if index < len(pca_df):
        fig.add_trace(
            go.Scatter3d(
                x=[pca_df.iloc[index]['PC1']],
                y=[pca_df.iloc[index]['PC2']],
                z=[pca_df.iloc[index]['PC3']],
                mode='markers',
                marker=dict(
                    size=10,
                    #symbol='x',
                    color=marker_color,
                    line=dict(color='black', width=1)
                ),
                name=f"Predicted Type: {predicted_type}"
            )
        )
    else:
        print(f"Invalid index: {index}")

# Display the plot
fig.show()
````
