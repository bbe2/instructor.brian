### **`8.2.2 - Advanced iteration styles`**


```
 recursion
# • idea: a function solves a problem by calling a smaller version of itself
# • needs a base case (smallest input you can answer directly) so it stops
# • recipe: if base → return simple value; else → return f(smaller input) combined
# • example math: factorial(n) = n × factorial(n-1), with factorial(0) = 1
```



````python

# concept demo: base case + one smaller self-call
def factorial(n: int) -> int:
    if n == 0:                 # base case
        return 1
    return n * factorial(n-1)  # recursive step (one smaller call)

# quick check
print("4! =", factorial(4))  # 24
````

````python
# generator example: yields one value at a time → memory-efficient iteration
# now with a basic phone keypad, starting from the key '8' and stopping at '0'

# keypad data (flattened)
pad = ['1','2','3','4','5','6','7','8','9','*','0','#']

def counter(n: int, start: int):
    """yield index positions, one by one"""
    for i in range(n):
        yield start + i

# start at '8'
start_idx = pad.index('3')

# loop OUTSIDE the function: map indices to keys and stop at '0'
for idx in counter(n=len(pad), start=start_idx):
    key = pad[idx % len(pad)]
    print(key)
    if key == '0':
        print("found 0 — stop")
        break
````
