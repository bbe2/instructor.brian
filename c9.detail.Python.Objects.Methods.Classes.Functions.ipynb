> perform behaviors in a class such as data arrangement, computation, printing, and conditional logic trees, perhaps to test, parse, or look for specific information

---

> For example, an object's unique ID, color, name, or numeric value for use in a calculation

---

**`Summary overview`**

``````````````````````````````````````python

 [section 9.1 overview — ## **`9.1 Definition: class, objects, methods, functions`**
by: Brian Hogan, brian.hogan@regiscollege.edu

In Python, a *class* is a blueprint that bundles data (attributes) and behavior
(methods) into a single template.

An *object* is a concrete instance of a class, carrying its own attribute
values, while *methods* are functions defined inside the class that
automatically receive the object (`self`) so they can read and update its
state.

Regular *functions* live outside classes and coordinate work between multiple
objects or perform general tasks, but conceptually they are the same kind of
callable building block as methods—just not bound to a specific instance.

▼
─> takeaway ➤ class = blueprint, object = instance, methods = functions bound
to that instance; functions outside the class orchestrate and reuse logic
across the program

Keywords:
==========
Term    | Role in Python OOP            | Description
--------|-------------------------------|----------------------------------------------
Class   | Blueprint / type definition   | describes shared attributes and methods for many objects
Object  | Instance of a class           | concrete “thing” with its own state based on a class
Method  | Function bound to an object   | defined inside a class; first parameter is usually `self`
Function| Reusable callable unit        | named block of code; may be standalone or used inside classes

``````````````````````````````````````

``````````````````````````````````````python
# =======================================================
# part 9.2.1: create the class and functions
# =======================================================

class shakespeare_minion:
    # parent object
    def __init__(self, name, total_plays):
        self.name = name
        self.perform_work = 0
        self.total_plays_not_read = total_plays
        self.total_plays_read = 0
        self.num_plays_work_now = 0

    #function.1
    def how_much_work_master(self):   # function 1: ask user how much work to do
        self.perform_work = int(input("enter greater than 0 to run program => "))

        if self.perform_work <= 0:
            sys.exit()  # on/off switch

        self.num_plays_work_now = int(
            input("enter how many plays you will read today => "))

    #funtion.2
    def do_work_and_report_status(self):   # function 2: do work and report status
        # no work queued
        if self.num_plays_work_now <= 0:
            return (
                "master, i have "
                + str(self.total_plays_not_read)
                + " plays left to read and no more work today." )

        # do one unit of work
        self.total_plays_not_read -= 1
        self.total_plays_read += 1
        self.num_plays_work_now -= 1

        if self.num_plays_work_now == 0:   # finished this batch
            return (
                "outcome a: master! "
                + self.name
                + " is done. i finished this many plays today! = "
                + str(self.total_plays_read)  )

        return (    # still more reading of plays to do?
            "outcome b: master, i read "
            + str(self.total_plays_read)
            + " plays today and have "
            + str(self.num_plays_work_now)
            + " more plays left to finish in this."   )

# =======================================================
# part 9.2.2: create a child object and run the functions
# =======================================================

# child object
minion = shakespeare_minion("Romeo", len(df0))

# ask how much work
minion.how_much_work_master()

# =======================================================
# part 9.2.3 run the transactions defined in the class
# =======================================================

print(minion.do_work_and_report_status())
print("------------------------------------------------------------")
print("------------------------------------------------------------")
print("total plays not read =",minion.total_plays_not_read)
print("number plays i didnt finish = ",minion.num_plays_work_now)
print("number of total plays i didnt read at all =",minion.total_plays_read)
``````````````````````````````````````

``````````````````````````````````````python
'''all code combined'''

#======================================================================
'''Lab'''
# Part Ia:   Import libraries and source data
# Part Ib:   Import libraries and source data
# Part II:  Create an object with functions to track work
# Part III: Instantiate and object, chile of parent, and run it on data
# Part IV:  Run a report
#======================================================================

'''=> # Part Ia    '''
#============================================================
#=> # Part I:  Import libraries and source data
#=============================================
import pandas as pd                               #dataframe library
import numpy as np                                #numeric library
import matplotlib.pyplot as plt                   #visualization library
import os                                         #operating system library
import sys                                        # sys.exit()
      '''if importing from your directory'''
#os.chdir('c:\\Users\\17574\\Desktop\\data_it304') #microsoft uses 2\\
#df0 = pd.read_csv("shakes_corpus_v1.csv")   #ETL method 1
#df0 = pd.read_excel("shakes_corpus_v1.xlsx") #ETL method 2'

# GitHub URL of the raw Excel file
github_url = 'https://github.com/bbe2/instructor.brian/raw/it.304.fall.2023/data.shakespeare.xlsx'
df0 = pd.DataFrame()                              #explicitly set datafarme
df0 = pd.read_excel("data_shakes_corpus_v1.xlsx") #ETL method 2
df0.info()  # RangeIndex: 37 entries,0 to 36, 4 col=> all data u need to index

mydict = df0.to_dict()                            #df to dict
    '''mydict_shakespeare => {'title':{},'script':{},'type':{},'ID:{} }'''
len(df0) #37                            lenth is always veritical by default!

df0.info()
   #             RangeIndex: 37 entries, 0 to 36
   #             Data columns (total 3 columns):
   #              #   Column  Non-Null Count  Dtype
   #             ---  ------  --------------  -----
   #              0   title   37 non-null     object
   #              1   script  37 non-null     object
   #              2   type    37 non-null     object
   #              3   ID      37 non-null     int64
   #             dtypes: int64(1), object(3) memory usage: 1.3+ KB

print(type(df0))  #use type() to always see what an object is
df0.head()
    #                                   title  ...    type
    #             0  Alls Well That Ends Well  ...  Comedy
    #             1            As You Like It  ...  Comedy

#2.1 use pandas df.to_dict() to move data into dictionary object
mydict = df0.to_dict()
print(mydict.keys())    #['title', 'script', 'type', 'ID'])
type(mydict.keys())     # object itself is keys


#===============================================
'''=> # Part Ib - explore the data   '''
#=========================================


#2.2 understand what a dictionary and zip is doing
mylist_keys = list(zip(mydict.keys()))
mylist_keys    # [('title',), ('script',), ('type',), ('ID',)]

#Inspect huge data and then break into smaller chunks
mylist_values = list(zip(mydict.values())) #WOW huge !
#point - zip helpful but continue to learn more functions

mylist_values   #=======================> #MEGASAURUS
                     #  35: 'Tragedy',
                     # 36: 'Tragedy'},),
                     #{0: 1,
                     # 1: 2,
                     # 2: 3,
'''2.1'''
type(mylist_values) #=> [({...})],

'''======> packed as [({...})], =>list, tuple, dictionary'''

#=====================================================
'''=> # Part Ic - manipulate data into different data objects  '''
#=============================================
'''Fucntion ideas & examples:
   i) write a function to count total characters in a play or all plays!
   ii) use an iterator, count characters, and put in a list
   iii) use new lists to create a report or write back to excel using'''

mylist = []                 #so this could be a function to count characters
for i in mydict['title'].values(): mylist.append(len(i))
print(mylist, type(mylist), sum(mylist))

#use the new objects and variables to creat a dictionary
myNewDict = {sum(mylist): mylist}
        #   or {'play-1':[<TitleTotalWords>,<ScriptTotalWords>]}

print(myNewDict)
print(type(myNewDict))
myDF = pd.DataFrame.from_dict(myNewDict)#function create a pandas.DF from dict
myDF.info()                               #check it out

''' Send to excel or view here - will review in class'''
#mywriter = pd.ExcelWriter('myoutput.xlsx') #create object that writes out
#myDF.to_excel(mywriter)
#mywriter.save()
#myDF                                #Excel will look exact same !



#=> Objective: use the following Classes example to make one of your own
#=> new function input("<message>") -> asks user for a value


#============================================================
'''=> # Part II    Draft an object with couple functions'''
    # We are training with .self notation. write self.<attribute or variable>
    # are inherent, or part of our instantiated children objects
#============================================================


class shakespeare_minion:           #this defines the parent object
  pass
  name = ""
  perform_work = 0                #yes,no switch so could exit terminal
  total_plays_not_read = len(df0) #use an object vs. hardcode a value
  total_plays_read = 0            #increment so you know how much work done
  num_plays_work_now = 0          # countdown tracker based on user input

  '''Function-1: ask user how many plays to read'''
  def how_much_work_master(self):
    perform_work = int(input("Enter greater than 0 to run program => "))
                  #int() function above ensures user input is encoded as number
    if perform_work <= 0:
      sys.exit()       #On/off switch so can exit program in terminal

    if perform_work > 0:
      self.num_plays_work_now = \
      int(input("Enter how many plays you will read today?=> "))
          #input function creates a display box
    perform_work = 0    #used 1x so destroying variable=good prorgramming

    '''Function-2: have minions completed what they said they would do?'''
    def do_work_and_report_status(self):
      #0) for transactions, here would be some kind of wait time to do work

      #1) condition 1 - Did we complete total work yet?
      if self.num_plays_work_now <= 0:
        #after test, then increment/decrement associated variables
        self.total_plays_not_read = self.total_plays_not_read - 1
        self.num_plays_work_now = self.num_plays_work_now - 1
        total_plays_read =+1 #another way to increment variables
        return "Master! {} is done. I finished {} plays today.". \
                            format(self.name,self.total_plays_read)

      #2) condition 2 - Still doing daily work ?
      elif self.num_plays_work_now > 0:
        #after test, then increment/decrement associated variables
        self.total_plays_not_read = self.total_plays_not_read - 1
        self.total_plays_read = self.total_plays_read +1
        self.num_plays_work_now = self.num_plays_work_now -1
        total_plays_read =+1        #another way to increment variables

      #3) condition 3 - this is a NESTED loop b/c now you either no more work
        #       or you report what you have left to do in this batch
      if self.num_plays_work_now == 0:
        return "Master I have {} plays left to read AND no more work.\
                I am 100% done for today so start over!".\
                format(self.total_plays_not_read)

      else:
        return "Master I read {} plays today and have {} more plays \
                to do in this most egregiousness and unjust batch.".\
                format(self.total_plays_read,self.num_plays_work_now)
  '''end of class & functional programming'''

''' =======================================================
    Part IIIa: Create a child object and run the functions
    ======================================================='''

# IIIa: ask user number plays to ready & run the transaction
'''Run these 3 lines together! - This starts to queue up total work'''
minion = shakespeare_minion()
minion.name = "Toothless Harold"
minion.how_much_work_master()         #ask user how much to do!

''' =======================================================
    Part IIIb: Run the transactions defined in the class
    ======================================================='''


#================= select all 4 lines - keep running to run out of work!
print(minion.do_work_and_report_status())
print(minion.total_plays_not_read)
print(minion.num_plays_work_now)
print(minion.total_plays_read)

#============================================================
'''#=> Part IIIc: Activity - write a little report '''
#============================================================
#trial and error - cut ok code
'''myplays = []   #get a minion name
for i in mydict['title'].values():
    myplays.append(i)
#normally no hardcoding! plays 23-33 have single names
minion_name = myplays[random.randint(23,33)]'''


#============================================================
#=> Housekeeping !
#============================================================
del myDF; del mydict; del mylist; del perform_work; del i


``````````````````````````````````````
